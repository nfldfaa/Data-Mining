```
cleaned = df.copy()
```

```
cleaned.columns = [str(x).strip().lower().replace(' ','_') for x in cleaned.columns]
```

```
for col in cleaned.columns:
    if any(k in col for k in ['tanggal','date','tgl']):
        cleaned[col] = pd.to_datetime(cleaned[col], errors='coerce')
```

```
num_cols_clean = cleaned.select_dtypes(include=[np.number]).columns.tolist()
cat_cols_clean = cleaned.select_dtypes(include=['object','category']).columns.tolist()
impute_info = {}
for c in num_cols_clean:
    if cleaned[c].isna().sum() > 0:
        med = cleaned[c].median()
        cleaned[c] = cleaned[c].fillna(med)
        impute_info[c] = ('median', med)
for c in cat_cols_clean:
    if cleaned[c].isna().sum() > 0:
        try:
            mode = cleaned[c].mode(dropna=True)[0]
            cleaned[c] = cleaned[c].fillna(mode)
            impute_info[c] = ('mode', mode)
        except:
            cleaned[c] = cleaned[c].fillna('missing')
            impute_info[c] = ('filled_missing','missing')

print("Imputation summary:", impute_info)
```

```
herd_col_candidates = [c for c in cleaned.columns if any(k in c for k in ['herd','jumlah','sapi','cattle','stock','populasi','population','total'])]
herd_col = herd_col_candidates[0] if herd_col_candidates else None
print("Herd column candidate:", herd_col)
if herd_col and np.issubdtype(cleaned[herd_col].dtype, np.number):
    cleaned['herd_size_bin'] = pd.cut(cleaned[herd_col], bins=[-1,5,20,50,200,1e9], labels=['very_small','small','medium','large','very_large'])
```

```
for c in num_cols_clean:
    try:
        if (cleaned[c] > 0).sum() / max(1, len(cleaned)) > 0.5 and cleaned[c].skew() > 1:
            cleaned[f'log_{c}'] = np.log1p(cleaned[c])
    except:
        pass
```

```
ohe_cols = [c for c in cat_cols_clean if cleaned[c].nunique() <= 10 and cleaned[c].nunique() > 1]
clean_for_ml = pd.get_dummies(cleaned, columns=ohe_cols, drop_first=True)
```

```
clean_csv_path = os.path.join(OUTDIR, "cleaned_data.csv")
clean_xlsx_path = os.path.join(OUTDIR, "cleaned_data.xlsx")
clean_for_ml.to_csv(clean_csv_path, index=False)
clean_for_ml.to_excel(clean_xlsx_path, index=False)
print("Saved cleaned data to:", clean_csv_path, clean_xlsx_path)

display(clean_for_ml.head())
```
